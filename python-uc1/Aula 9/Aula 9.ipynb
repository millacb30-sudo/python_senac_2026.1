```python
produtos = (
    ("Arroz", 5.99, 10 ),
    ("Feijão", 7.49,  3),
    ("Leite", 4.89,  10),
    ("Óleo", 9.99,  4),
    ("Açúcar", 3.29,  5)
)
valor_total=0
valor_produto=0
print("\nLISTA DE PRODUTOS")

valor_total+=valor_produto
print("LISTA DE PRODUTOS")
for nome, preco, qt in produtos:
    valor_produto=preco* qt
    valor_total+=valor_produto
    print(f"{nome:.<20} R$ {preco:6.2f} QT {qt:03d} .. R$ {valor_produto:8.2f}")

print(f"\n\t Valor Total da Compra: R$ {valor_total:7.2f}")
```
