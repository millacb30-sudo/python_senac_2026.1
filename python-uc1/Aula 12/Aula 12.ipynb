```python
chaves = ["nome", "dt_nascimento", "cep", "logradouro", "bairro", "cidade", "estado"]
valor_padrao = "N/A"
usuario = dict.fromkeys(chaves,valor_padrao)

print(usuario)
```

```python
usuario["nome"] = input("Digite o seu nome:")
```

```python
print(usuario["nome"])
```

```python
usuario["nome"] = input("Digite o seu nome:")
usuario["dt_nascimento"] = input("Digite sua data de nascimento:")
usuario["cep"] = input("Digite o seu cep:")
usuario["logradouro"] = input("Digite o logradouro:")
usuario["bairro"] = input("Digite o bairro:")
usuario["cidade"] = input("Digite a cidade:")
usuario["estado"] = input("Digite o seu estado:")
```

```python
print(usuario)
```

```python
for chave, valor in usuario.items():
    print(f"{chave.capitalize():_<20}: {valor.capitalize()}")
```
